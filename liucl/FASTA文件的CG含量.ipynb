```python
import gzip
```

```python
def calculate_gc_content(sequence):
    sequence = sequence.upper()
    gc = sequence.count("G") + sequence.count("C")
    total = len(sequence)
    return (gc / total * 100) if total > 0 else 0
```

```python
# 初始化统计变量
total_gc = 0          # 全部 GC 数
total_bases = 0       # 全部碱基数
lower_gc = 0          # 小写部分 GC 数
lower_bases = 0       # 小写部分碱基数
upper_gc = 0          # 大写部分 GC 数
upper_bases = 0       # 大写部分碱基数
```

```python
# 存储每条序列的详细信息
with gzip.open("F:\\病原体\\培训\\Phialophora_verrucosa\\GCA_016109465.1.fna.gz", "rt") as f:
    current_header = ""
    current_sequence = []

    for line in f:
        line = line.strip()

        if line.startswith(">"):  # 如果是头信息行
            if current_sequence:  # 处理前一条序列
                full_seq = "".join(current_sequence)
                seq_length = len(full_seq)
                gc_content = calculate_gc_content(full_seq)

                # 记录当前序列的详细信息
                sequence_info.append({
                    "header": current_header,
                    "length": seq_length,
                    "gc_content": gc_content
                })

                # 更新全局统计
                total_gc += (full_seq.upper().count("G") + full_seq.upper().count("C"))
                total_bases += seq_length

                # 区分大小写统计（如果需要）
                lower_seq = "".join([c for c in full_seq if c.islower()])
                upper_seq = "".join([c for c in full_seq if c.isupper()])

                lower_gc += (lower_seq.upper().count("G") + lower_seq.upper().count("C"))
                lower_bases += len(lower_seq)
                upper_gc += (upper_seq.upper().count("G") + upper_seq.upper().count("C"))
                upper_bases += len(upper_seq)

            # 重置为新的序列
            current_header = line[1:]  # 去掉 ">"
            current_sequence = []

        else:  # 如果是序列行
            current_sequence.append(line)

    # 处理最后一条序列
    if current_sequence:
        full_seq = "".join(current_sequence)
        seq_length = len(full_seq)
        gc_content = calculate_gc_content(full_seq)

        sequence_info.append({
            "header": current_header,
            "length": seq_length,
            "gc_content": gc_content
        })

        # 更新全局统计
        total_gc += (full_seq.upper().count("G") + full_seq.upper().count("C"))
        total_bases += seq_length

        lower_seq = "".join([c for c in full_seq if c.islower()])
        upper_seq = "".join([c for c in full_seq if c.isupper()])

        lower_gc += (lower_seq.upper().count("G") + lower_seq.upper().count("C"))
        lower_bases += len(lower_seq)
        upper_gc += (upper_seq.upper().count("G") + upper_seq.upper().count("C"))
        upper_bases += len(upper_seq)
```

```python
# 计算全局 GC 含量
total_gc_content = (total_gc / total_bases * 100) if total_bases > 0 else 0
lower_gc_content = (lower_gc / lower_bases * 100) if lower_bases > 0 else 0
upper_gc_content = (upper_gc / upper_bases * 100) if upper_bases > 0 else 0

# 输出每条序列的详细信息
print("\n=== 每条序列的 GC 含量 ===")
for seq in sequence_info:
    print(f"序列头: {seq['header']}")
    print(f"长度: {seq['length']} bp")
    print(f"GC 含量: {seq['gc_content']:.2f}%")
    print("-" * 50)

# 输出全局统计结果
print("\n=== 全局 GC 含量统计 ===")
print(f"全部序列 GC 含量: {total_gc_content:.2f}%")
print(f"小写部分 GC 含量: {lower_gc_content:.2f}%")
print(f"大写部分 GC 含量: {upper_gc_content:.2f}%")
print(f"总碱基数: {total_bases} bp")
```
